## Round 1：多資料對比（兩條線）

```python
# Round 1 TASK:
# 1) 生成兩條數列：學習時間(hours)、成績(score)
# 2) 畫兩條線比較走勢（同 X 軸）
# 3) 練習：資料對比、label、legend

import matplotlib.pyplot as plt
import random

random.seed(2)

weeks = list(range(1, 11))
study_hours = [random.randint(3, 12) for _ in weeks]
scores = [min(100, 60 + h*3 + random.randint(-5, 5)) for h in study_hours]

plt.figure()
plt.plot(weeks, study_hours, marker="o", label="Study Hours")
plt.plot(weeks, scores, marker="o", label="Score")
plt.title("Compare Two Series")
plt.xlabel("Week")
plt.legend()
plt.show()
```

## Round 2：用圖表說故事（自動生成結論）

```python
# Round 2 TASK:
# 1) 根據 scores 的趨勢，自動判斷「整體上升/下降/平穩」
# 2) 印出一句圖表故事（data storytelling）
# 3) 練習：簡單統計 + if 判斷

import numpy as np

scores = np.array([65, 68, 70, 73, 75, 78, 80, 82, 83, 85])

diff = scores[-1] - scores[0]

print("scores:", scores.tolist())
if diff > 8:
    print("📈 故事：整體分數明顯上升，學習策略可能有效。")
elif diff < -8:
    print("📉 故事：整體分數下降，可能需要調整學習方法。")
else:
    print("➖ 故事：整體變化不大，可能需要更長期觀察或加入新方法。")
```
