## Round 1：趨勢線概念（用 numpy 手刻「找一條線」）

```python
# Round 1 TASK:
# 1) 給你 x=學習時數, y=分數（小資料）
# 2) 用 numpy.polyfit 找出趨勢線 y = ax + b
# 3) 畫散點圖 + 趨勢線

import numpy as np
import matplotlib.pyplot as plt

x = np.array([1, 2, 3, 4, 5, 6])
y = np.array([55, 60, 62, 70, 72, 78])

a, b = np.polyfit(x, y, 1)  # 一次式
print(f"趨勢線：y = {a:.2f}x + {b:.2f}")

y_hat = a*x + b

plt.figure()
plt.scatter(x, y, label="data")
plt.plot(x, y_hat, label="trend line")
plt.title("Linear Trend (Beginner Regression)")
plt.xlabel("Study Hours")
plt.ylabel("Score")
plt.legend()
plt.show()
```

## Round 2：預測小工具（輸入學習時數 -> 預測分數）

```python
# Round 2 TASK:
# 1) 先用 polyfit 得到 a, b
# 2) 讓使用者輸入學習時數 hours
# 3) 輸出預測分數（並做防呆）

import numpy as np

x = np.array([1, 2, 3, 4, 5, 6])
y = np.array([55, 60, 62, 70, 72, 78])
a, b = np.polyfit(x, y, 1)

def safe_float_input(prompt):
    while True:
        s = input(prompt).strip()
        try:
            return float(s)
        except:
            print("⚠️ 請輸入數字（可以有小數）。")

hours = safe_float_input("你這週預計讀書幾小時？ ")
pred = a*hours + b
pred = max(0, min(100, pred))

print(f"預測分數：約 {pred:.1f} 分")
```
