## Round 1：圖片 = 像素矩陣（用 numpy 建立假圖 + 顯示）

```python
# Round 1 TASK:
# 1) 用 numpy 生成一張 28x28 的灰階圖片（0~255）
# 2) 顯示這張圖
# 3) 讓學生理解：圖片就是矩陣

import numpy as np
import matplotlib.pyplot as plt

np.random.seed(0)

img = np.random.randint(0, 256, size=(28, 28), dtype=np.uint8)

plt.figure()
plt.imshow(img, cmap="gray")
plt.title("A Random 28x28 Image (Pixel Matrix)")
plt.axis("off")
plt.show()

print("img shape:", img.shape)
print("top-left 5x5 pixels:\n", img[:5, :5])
```

## Round 2：簡易影像分類（用規則做「亮/暗」分類）

```python
# Round 2 TASK:
# 1) 讀入/生成多張 28x28 圖
# 2) 用平均亮度當特徵
# 3) 規則分類：avg > threshold -> 'bright' else 'dark'
# 4) 印出準確率（toy example）

import numpy as np

np.random.seed(1)

def make_image(label):
    # bright: 偏亮；dark: 偏暗
    if label == "bright":
        return np.random.randint(120, 256, size=(28,28))
    else:
        return np.random.randint(0, 136, size=(28,28))

# dataset
labels = ["bright"]*50 + ["dark"]*50
images = [make_image(lbl) for lbl in labels]

# classify
threshold = 128
preds = []
for img in images:
    avg = img.mean()
    pred = "bright" if avg > threshold else "dark"
    preds.append(pred)

acc = sum(p==y for p,y in zip(preds, labels)) / len(labels)
print("threshold:", threshold)
print("accuracy:", acc)
```
