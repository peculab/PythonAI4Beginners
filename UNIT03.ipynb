# 2026/03/20 — 檔案與資料處理

## Round 1：讀寫檔案（簡易日記 / ToDo）

```python
# Round 1 TASK:
# 1) 讓使用者輸入一段文字（今天的紀錄）
# 2) 寫進檔案 notes.txt（append）
# 3) 再把 notes.txt 全部讀出來顯示

filename = "notes.txt"

text = input("輸入今天的一句話（日記/ToDo）：").strip()

with open(filename, "a", encoding="utf-8") as f:
    f.write(text + "\n")

print("\n✅ 已寫入 notes.txt。")
print("\n=== notes.txt 內容 ===")
with open(filename, "r", encoding="utf-8") as f:
    print(f.read())
```

## Round 2：pandas CSV 清理（缺值、欄位型別、簡單統計）

```python
# Round 2 TASK:
# 1) 建立一份小 CSV（在 Colab 直接生成）
# 2) 用 pandas 讀入
# 3) 清理：缺值補平均、把字串數字轉成數值
# 4) 輸出平均、最大、最小

!pip -q install pandas

import pandas as pd
import numpy as np

# 生成示例 CSV
data = {
    "name": ["Amy", "Ben", "Cathy", "David"],
    "score": ["90", "78", "", "95"]  # Cathy 缺值（空字串）
}
df = pd.DataFrame(data)
csv_path = "scores.csv"
df.to_csv(csv_path, index=False)

# 讀入
df = pd.read_csv(csv_path)

# 把 score 轉成數值（不能轉的會變 NaN）
df["score"] = pd.to_numeric(df["score"], errors="coerce")

# 缺值用平均補
mean_score = df["score"].mean()
df["score"] = df["score"].fillna(mean_score)

print("=== 清理後資料 ===")
print(df)

print("\n=== 統計 ===")
print("平均：", df["score"].mean())
print("最大：", df["score"].max())
print("最小：", df["score"].min())
```
