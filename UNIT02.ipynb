## Round 1：List / Dict 做小系統（簡易「點餐系統」）

```python
# Round 1 TASK:
# 1) 用 dict 表示菜單：品項 -> 價格
# 2) 讓使用者一直輸入品項直到輸入 'done'
# 3) 用 list 存購物車，最後輸出總金額與明細

menu = {
    "burger": 120,
    "cola": 30,
    "coffee": 60,
    "salad": 80
}

cart = []  # 放使用者點的品項（字串）

print("=== 點餐系統 ===")
print("菜單：", menu)
print("輸入品項名稱加入購物車，輸入 done 結帳。")

while True:
    item = input("你要點什麼？ ").strip().lower()
    if item == "done":
        break
    if item not in menu:
        print("沒有這個品項，請重試。")
        continue
    cart.append(item)
    print(f"已加入：{item}")

total = 0
print("\n=== 結帳 ===")
for item in cart:
    price = menu[item]
    total += price
    print(f"- {item}: {price} 元")

print("總金額：", total, "元")
```

## Round 2：排序 / 排行榜設計（成績排行榜）

```python
# Round 2 TASK:
# 1) 用 dict 存學生 -> 分數
# 2) 把它轉成 list of tuples 並排序（分數由高到低）
# 3) 印出 Top 3 排行榜

scores = {
    "Amy": 91,
    "Ben": 78,
    "Cathy": 88,
    "David": 95,
    "Ella": 83
}

items = list(scores.items())  # [('Amy', 91), ...]
items.sort(key=lambda x: x[1], reverse=True)

print("=== 成績排行榜 ===")
for i, (name, score) in enumerate(items, start=1):
    print(f"{i}. {name} - {score}")

print("\n=== Top 3 ===")
top3 = items[:3]
for i, (name, score) in enumerate(top3, start=1):
    print(f"🥇🥈🥉"[i-1], name, score)
```
