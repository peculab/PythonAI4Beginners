## Round 1：折線圖看趨勢

```python
# Round 1 TASK:
# 1) 生成一個 30 天的「心情分數」(0~100)
# 2) 畫折線圖看趨勢
# 3) 練習：list、迴圈、matplotlib

import random
import matplotlib.pyplot as plt

random.seed(1)

days = list(range(1, 31))
mood = []
now = 60
for _ in days:
    now += random.randint(-5, 5)
    now = max(0, min(100, now))
    mood.append(now)

plt.figure()
plt.plot(days, mood, marker="o")
plt.title("Mood Trend (30 days)")
plt.xlabel("Day")
plt.ylabel("Mood Score")
plt.show()
```

## Round 2：長條圖與比較（兩組班級平均）

```python
# Round 2 TASK:
# 1) 建立兩個班級的平均分數
# 2) 用長條圖比較
# 3) 練習：字典、bar chart

import matplotlib.pyplot as plt

classes = ["Class A", "Class B"]
avg_scores = [82, 89]

plt.figure()
plt.bar(classes, avg_scores)
plt.title("Average Score Comparison")
plt.xlabel("Class")
plt.ylabel("Average Score")
plt.show()
```
