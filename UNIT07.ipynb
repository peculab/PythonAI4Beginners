## Round 1：規則式對話機器人（if/elif）

```python
# Round 1 TASK:
# 1) 做一個簡單聊天機器人
# 2) 支援：問候、查課表、離開
# 3) 練習：while、字串處理、if/elif

print("=== Rule-based Chatbot ===")
print("輸入 hi / schedule / bye")

while True:
    msg = input("你：").strip().lower()
    if msg in ["hi", "hello"]:
        print("機器人：嗨！今天想學什麼？")
    elif msg == "schedule":
        print("機器人：本週主題是聊天機器人！")
    elif msg in ["bye", "exit", "quit"]:
        print("機器人：掰掰～下次見！")
        break
    else:
        print("機器人：我還聽不懂，你可以輸入 hi / schedule / bye")
```

## Round 2：對話流程設計（狀態機：選單式流程）

```python
# Round 2 TASK:
# 1) 加入「狀態 state」：menu / ordering
# 2) 在 ordering 狀態下可以點餐，done 結帳回 menu
# 3) 練習：狀態機概念

menu = {"burger": 120, "cola": 30}
cart = []
state = "menu"

print("=== Chatbot with State ===")
print("指令：order / help / bye")

while state != "bye":
    msg = input("你：").strip().lower()

    if state == "menu":
        if msg == "help":
            print("機器人：輸入 order 進入點餐；輸入 bye 離開。")
        elif msg == "order":
            state = "ordering"
            print("機器人：進入點餐模式。可點 burger / cola，輸入 done 結帳。")
        elif msg == "bye":
            print("機器人：掰掰！")
            break
        else:
            print("機器人：menu 狀態只懂 help / order / bye")

    elif state == "ordering":
        if msg == "done":
            total = sum(menu[i] for i in cart)
            print("機器人：你點了", cart, "總共", total, "元")
            cart = []
            state = "menu"
            print("機器人：回到 menu。")
        elif msg in menu:
            cart.append(msg)
            print("機器人：已加入", msg)
        else:
            print("機器人：點餐只接受 burger / cola / done")
```
