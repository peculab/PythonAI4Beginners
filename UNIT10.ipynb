## Round 1：抓網路資料（requests + BeautifulSoup，抓標題）

```python
# Round 1 TASK:
# 1) 用 requests 抓一個網頁 HTML
# 2) 用 BeautifulSoup 解析並抓出 title
# 3) 練習：HTTP、解析、例外處理
#
# NOTE: 某些網站會擋爬蟲。示範用 example.com（穩定）

!pip -q install requests beautifulsoup4

import requests
from bs4 import BeautifulSoup

url = "https://www.ptt.cc/bbs/Japan_Travel/index.html"

resp = requests.get(url, timeout=10)
resp.raise_for_status()

soup = BeautifulSoup(resp.text, "html.parser")
print("Title:", soup.title.text.strip())
```

## Round 2：情緒分析 + 小專題（超簡單情緒詞典法）

```python
# Round 2 TASK:
# 1) 先做一個超簡單情緒詞典 positive/negative
# 2) 對抓到的文字（或自己輸入的句子）做情緒分數
# 3) 給出結論：positive / negative / neutral

positive_words = {"good", "great", "happy", "love", "excellent", "awesome"}
negative_words = {"bad", "sad", "hate", "terrible", "awful", "angry"}

def sentiment_score(text):
    tokens = [t.strip(".,!?;:").lower() for t in text.split()]
    pos = sum(t in positive_words for t in tokens)
    neg = sum(t in negative_words for t in tokens)
    return pos - neg, pos, neg

text = input("輸入一句英文（做情緒分析）：\n")
score, pos, neg = sentiment_score(text)

print("pos:", pos, "neg:", neg, "score:", score)

if score > 0:
    print("Result: POSITIVE 🙂")
elif score < 0:
    print("Result: NEGATIVE 🙁")
else:
    print("Result: NEUTRAL 😐")
```
