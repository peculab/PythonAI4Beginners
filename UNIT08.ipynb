## Round 1：LLM / Prompt 概念（先用「假 LLM」理解 prompt）

```python
# Round 1 TASK:
# 1) 用一個 "fake_llm(prompt)" 模擬 LLM：根據 prompt 回答
# 2) 讓學生理解：prompt = 你如何描述任務與角色
# 3) 不用 API（初學者友善）

def fake_llm(prompt):
    prompt_lower = prompt.lower()
    if "summarize" in prompt_lower:
        return "這是一段摘要（示範）：重點是把長內容變短。"
    if "translate" in prompt_lower and "chinese" in prompt_lower:
        return "這是一段翻譯（示範）：我把英文翻成中文。"
    if "teacher" in prompt_lower:
        return "好的，我用老師口吻一步一步說明。"
    return "我收到你的 prompt 了（示範回覆）。"

prompt = input("輸入你的 prompt：\n")
print("\n=== fake LLM response ===")
print(fake_llm(prompt))
```

## Round 2：AI 助手設計（模板化 prompt + 工具函數）

```python
# Round 2 TASK:
# 1) 建立 prompt template：角色 + 任務 + 輸入資料
# 2) 寫一個 build_prompt(role, task, user_text)
# 3) 把 prompt 丟給 fake_llm（先理解結構）

def build_prompt(role, task, user_text):
    return f"""ROLE: {role}
TASK: {task}
INPUT:
{user_text}
OUTPUT:"""

def fake_llm(prompt):
    # 簡化版：只用關鍵字示範
    if "ROLE: Teacher" in prompt:
        return "（示範）老師口吻：先講概念，再給例子，最後練習。"
    return "（示範）一般回答。"

role = "Teacher"
task = "Explain Python functions in beginner-friendly way."
user_text = input("你想問 AI 什麼？\n")

prompt = build_prompt(role, task, user_text)

print("\n=== Prompt ===")
print(prompt)
print("\n=== AI response ===")
print(fake_llm(prompt))
```
